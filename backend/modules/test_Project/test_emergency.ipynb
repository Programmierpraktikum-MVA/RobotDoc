```python
# Importiere den Emergency Detector
import sys
import os

# Hole den Pfad des aktuellen Notebooks
project_root = os.getcwd()  
modules_path = os.path.join(project_root, '..') 
modules_path = os.path.abspath(modules_path)

if modules_path not in sys.path:
    sys.path.insert(0, modules_path)

from MedicalAnalysis.emergency_detection import detect_emergency_extended, get_emergency_response_extended

# Teste verschiedene Nachrichten

test_messages_english = [
    "I have severe chest pain and difficulty breathing",
    "My child has high fever",
    "I have a headache",
    "I cannot breathe and feel dizzy",
    "I have a mild cold",
    "I have a stroke - cannot speak anymore",
    "I have severe burns on my arm",
    "I have sudden vision changes",
    "I have a severe allergic reaction",
    "I just have a little cough",
    "I am unconscious and cannot get up",
    "I have breathing difficulties and feel very bad",
    "I have sudden speech problems",
    "I have a seizure"
]

print("=== Original Emergency Detector Test (English) ===\n")

for i, message in enumerate(test_messages_english, 1):
    print(f"Test {i}: {message}")
    
    # Verwende den ursprünglichen Detector
    from MedicalAnalysis.emergency_detection import detect_emergency_extended, get_emergency_response_extended
    
    is_emergency, priority, symptoms = detect_emergency_extended(message)
    
    print(f"  Emergency: {is_emergency}")
    print(f"  Priority: {priority}")
    print(f"  Symptoms: {symptoms}")
    
    if is_emergency:
        response = get_emergency_response_extended(priority, symptoms)
        print(f"  Response: {response}")
    
    print("-" * 60)
```
